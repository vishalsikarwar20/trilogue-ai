```python
import os
import requests
from dotenv import load_dotenv
from openai import OpenAI
from IPython.display import Markdown,display
```

```python
load_dotenv(override=True)
google_api_key=os.getenv('GOOGLE_API_KEY')
openrouter_api_key=os.getenv('OPENROUTER_API_KEY')
if google_api_key:
    print(f"this is the google api key {google_api_key[:4]}")
else:
    print("google api key is workking")    
if openrouter_api_key:
    print(f"this is the openrouter api key {openrouter_api_key[:4]}")    
else:
    print("openrouter api key is not working")    
```

```python
openrouter_url = "https://openrouter.ai/api/v1"
gemini_url = "https://generativelanguage.googleapis.com/v1beta/openai/"

gemini = OpenAI(api_key=google_api_key, base_url=gemini_url)
openrouter = OpenAI(base_url=openrouter_url, api_key=openrouter_api_key)
```

```python
nvidia_model = "nvidia/nemotron-3.5-lightning:free"
gemini_model = "gemini-3.1-flash-lite"
alex_model="nvidia/nemotron-3.5-lightning:free"

nvidia_system = " you are a blake , a chatbot who is very argumentative; \
you disagree with anything in the conversation and you challenge everything, in a snarky way."

gemini_system = "You are a charlie, very polite, courteous chatbot. You try to agree with \
everything the other person says, or find common ground. If the other person is argumentative, \
you try to calm them down and keep chatting."

alex_system=""" 
your are a alex , a chatbot who is very argumentative; you disagree with anythin in the conversation \
 and  you challenge everything, a snarky way. you are in the conversation with blake and charlie.
    """


conversation=[("blake","oii u"),("charlie","hyee"),("alex","helloo")]



```

```python
def call_nvidia():
    messages=[{"role":"system","content":nvidia_system}]
    chat=""
    for name, mssg in conversation:
        chat+=f"{name}: \n {mssg}"
        blake_prompt=f"""you are blake ,in the conversation with charlie and alex. the conversation so far is as follows: {chat} \
        now with this, respond with what you would like to say next, as blake."""
    messages.append({"role":"user","content":blake_prompt})
        

    response=openrouter.chat.completions.create(
        model=nvidia_model,
        messages=messages
    )
    # if not response.choices:
    #     print("RAW:", response)
    #     print("EXTRA:", response.model_extra)
    #     return None
       
    return response.choices[0].message.content
```

```python
def call_gemini():
    messages=[{"role":"system","content":gemini_system}]
    chat=""
    for name,mssg in (conversation):
        chat+=f"{name}: \n {mssg}"
        charlie_prompt=f"""you are charlie ,in the conversation with blake and alex. the conversation so far is as follows: {chat} \
        now with this, respond with what you would like to say next, as charlie."""
        
    messages.append({"role":"user","content":charlie_prompt})

    response=gemini.chat.completions.create(
        model=gemini_model,
        messages=messages
        )
    return response.choices[0].message.content    
```

```python
def call_alex():
    messages=[{"role":"system","content":alex_system}]
    chat=""
    for name,mssg in conversation:
        chat+=f"{name}: \n {mssg}"
        alex_prompt=f"""you are alex ,in the conversation with blake and charlie. the conversation so far is as follows: {chat} \
        now with this, respond with what you would like to say next, as alex."""
        
    messages.append({"role":"user","content":alex_prompt})

    response=openrouter.chat.completions.create(
        model=alex_model,
        messages=messages
        )
    return response.choices[0].message.content    
```

```python
call_nvidia()
```

```python
call_gemini()
```

```python
call_alex()
```

```python
for name,mssg in conversation:
    display(Markdown(f"{name}: \n {mssg}\n"))

for i in range(2):
    nvidia_next=call_nvidia()
    display(Markdown(f"nvidia boy-blake : {nvidia_next} \n "))
    conversation.append(("blake",nvidia_next))

    gemini_next=call_gemini()
    display(Markdown(f"gemini boy charlie : {gemini_next} \n "))
    conversation.append(("charlie",gemini_next))

    alex_next=call_alex()
    display(Markdown(f" nvidia boy alex : {alex_next} \n "))
    conversation.append(("alex",alex_next))

```
